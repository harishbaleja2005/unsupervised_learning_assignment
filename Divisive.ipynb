# Hierarchical Clustering on Iris Assignment

## 1. Load the Iris dataset using Python's sklearn, perform hierarchical clustering with the 'ward' linkage, and plot the resulting dendrogram using scipy's dendrogram() function.

### Answer

```python
from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import linkage,dendrogram
import matplotlib.pyplot as plt
iris=load_iris()
X=iris.data
Z=linkage(X,method='ward')
plt.figure(figsize=(10,5))
dendrogram(Z)
plt.title('Ward Linkage Dendrogram')
plt.show()
```

## 2. Visually inspect the dendrogram you plotted and cut it at a height that results in 3 distinct clusters. Assign each data point to its cluster and print the number of samples in each cluster.

### Answer

```python
from scipy.cluster.hierarchy import fcluster
import pandas as pd
clusters=fcluster(Z,t=15,criterion='distance')
print(pd.Series(clusters).value_counts().sort_index())
```

## 3. Repeat the dendrogram plotting for the same data using 'single' and 'complete' linkage methods, and compare how the cluster shapes and heights differ from 'ward' linkage.

### Answer

```python
fig,ax=plt.subplots(1,3,figsize=(18,5))
for a,m in zip(ax,['ward','single','complete']):
    dendrogram(linkage(X,method=m),ax=a,no_labels=True)
    a.set_title(m.capitalize())
plt.tight_layout()
plt.show()
```

## 4. Interpret the dendrogram you created.

### Answer

```python
print('Clusters that merge at lower heights are more similar.')
print('Clusters joining at higher heights are more different.')
print('Ward linkage forms compact, balanced clusters.')
print('Long vertical branches indicate greater separation between groups.')
```
