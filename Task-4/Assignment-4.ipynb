# PCA and KMeans Assignment

## 1. Load the Iris dataset using sklearn, apply PCA to reduce it to 2 principal components, and print the explained variance ratio for both components.

### Answer

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
iris=load_iris()
X=iris.data
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X)
print("Explained Variance Ratio:",pca.explained_variance_ratio_)
```

## 2. Plot the 2D projection of the Iris dataset after PCA transformation, using a different color for each species.

### Answer

```python
import matplotlib.pyplot as plt
colors=['red','green','blue']
for i,c in enumerate(colors):
    plt.scatter(X_pca[iris.target==i,0],X_pca[iris.target==i,1],label=iris.target_names[i])
plt.xlabel('PC1');plt.ylabel('PC2');plt.legend();plt.title('Iris PCA');plt.show()
```

## 3. Given a dataset of Zomato restaurant ratings with 10 features, use PCA from sklearn to reduce the data to 3 components and visualize the cumulative explained variance as a line plot.

### Answer

```python
import numpy as np
np.random.seed(42)
zomato=np.random.rand(500,10)
from sklearn.decomposition import PCA
pca3=PCA(n_components=3)
pca3.fit_transform(zomato)
cum=np.cumsum(pca3.explained_variance_ratio_)
plt.figure()
plt.plot(range(1,4),cum,marker='o')
plt.xlabel('Components');plt.ylabel('Cumulative Explained Variance')
plt.title('Zomato PCA')
plt.grid(True)
plt.show()
```

## 4. Take the PCA-transformed Iris data (2 components) and run KMeans clustering with 3 clusters. Plot the clusters and centroids on the 2D PCA plot.

### Answer

```python
from sklearn.cluster import KMeans
kmeans=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=kmeans.fit_predict(X_pca)
cent=kmeans.cluster_centers_
plt.figure()
plt.scatter(X_pca[:,0],X_pca[:,1],c=labels)
plt.scatter(cent[:,0],cent[:,1],marker='X',s=200)
plt.title('KMeans on PCA Iris')
plt.show()
```

## 5. Use the silhouette_score function from sklearn.metrics to evaluate the quality of your KMeans clusters on the PCA-reduced Iris data. Print the silhouette score and explain what it indicates about your clustering result in 2-3 lines.

### Answer

```python
from sklearn.metrics import silhouette_score
score=silhouette_score(X_pca,labels)
print("Silhouette Score:",score)
print("\nExplanation:")
print("A score close to 1 indicates well-separated clusters.")
print("A score around 0 indicates overlapping clusters.")
print("The obtained score suggests the clustering quality on the PCA-reduced Iris dataset.")
```
